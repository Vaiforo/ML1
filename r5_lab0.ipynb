### Инициализация проекта

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import warnings
from ucimlrepo import fetch_ucirepo 
```

```python
wine_quality = fetch_ucirepo(id=186) 
  
X = wine_quality.data.features 
y = wine_quality.data.targets
X
```

Несколько говорящих о хорошем вкусе параметров, которые можно сразу выделить:
- Низкая летучая кислотность
- Высокое сожержание лимонной кислоты
- Низкое содержание хлоридов
- Низкое содержание сульфатов
- Высокая крепость

Попробуем ещё подкрепить это:

```python
volatile_acidity = data.groupby(['quality']).quantile(0.5)
volatile_acidity
```

Результаты аналогичны

Но есть отличия в показания у свободного диоксида серы "free_sulfur_dioxide"

```python
volatile_acidity = (data.groupby(['quality']).quantile(0.25)['free_sulfur_dioxide'] + data.groupby(['quality']).quantile(0.75)['free_sulfur_dioxide']) / 2
volatile_acidity
```

Так и есть, чем выше показатель, тем выше оценивается вкус

## Анализ по графикам

### Фиксированная кислотность "fixed_acidity"

```python
sns.lmplot(x="quality", y="fixed_acidity", data=data, fit_reg=False)
plt.title('fixed_acidity')
```

### Летучая кислотность "volatile_acidity"

```python
sns.lmplot(x="quality", y="volatile_acidity", data=data, fit_reg=False)
plt.title('volatile_acidity')
```

### Лимонная кислота "citric_acid"

```python
sns.lmplot(x="quality", y="citric_acid", data=data, fit_reg=False)
plt.title('citric_acid')
```

### Остаточный сахар "residual_sugar"

```python
sns.lmplot(x="quality", y="residual_sugar", data=data, fit_reg=False)
plt.title('residual_sugar')
```

### Хлориды "chlorides"

```python
sns.lmplot(x="quality", y="chlorides", data=data, fit_reg=False)
plt.title('chlorides')
```

```python
sns.lmplot(x="quality", y="free_sulfur_dioxide", data=data, fit_reg=False)
plt.title('free_sulfur_dioxide')
```

```python
g = sns.JointGrid(data=data, x="volatile_acidity", y="quality")
g.plot(sns.kdeplot, sns.histplot)
plt.show()
```

```python
g = sns.FacetGrid(data, row="quality", hue="quality", aspect=3, height=1.5)
g.map(sns.kdeplot, "volatile_acidity", fill=True, alpha=0.7)

g.set(yticks=[], ylabel="")

plt.show()
```

```python
corr_matrix = data.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5, cbar=True)

plt.title("Correlation Heatmap")
plt.show()
```

```python
g = sns.FacetGrid(data, col="quality", height=4, aspect=1.2)

# Добавляем график KDE (оценка плотности)
g.map(sns.kdeplot, "volatile_acidity", fill=True, alpha=0.7)

# Добавляем заголовок
plt.subplots_adjust(top=0.8)
g.fig.suptitle("FacetGrid with KDE Projection for volatile_acidity by quality")

plt.show()
```
